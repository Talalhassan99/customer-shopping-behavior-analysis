```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
df['review_rating'] = df.groupby('category')['review_rating'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
#Feature Engineering
#Creating a new column age_group
labels = ['Young-adult','adult','middle-aged','senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age_group','age']].head(10)
```

```python
# new column frequency of puchase in number of days
frequency_mapping= {
        'Fortnightly': 14,
        'Weekly': 7,
        'Annually': 365,
        'Monthly': 30,
        'Bi-Weekly': 14,
        'Quarterly': 90,
        'Every 3 months': 90,
    }
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(20)
```

```python
df[['discount_applied','promo_code_used']].head(5)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop(columns='promo_code_used')
```

```python
import psycopg2
from sqlalchemy import create_engine

username = "postgres"
password = "root"
host = "localhost"
port = "5432"
database = "cust_behaviour"

engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

try:
    conn = psycopg2.connect(
        user=username,
        password=password,
        host=host,
        port=port,
        database=database
    )
    print("Connected successfully!")
    conn.close()
except Exception as e:
    print("Connection failed:", e)
```

```python
table_name = "customer"
df.to_sql(
    table_name,
    engine,
    if_exists='replace',
    index=False
)
print(f"data successfully loaded into the table '{table_name}' in database '{database}'.")
```
